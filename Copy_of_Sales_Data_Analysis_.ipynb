```
import pandas as pd
import matplotlib.pyplot as plt
data = {"product": ["laptop", "phone", "book", "laptop", "phone", "book"], "quantity": [2, 5, 10, 1, 3, 7], "price": [500, 300, 20,500,300,20]}
df = pd.DataFrame(data)
df["revenue"] = df["quantity"] * df["price"]
print(df)
```

```
total_revenue = df["revenue"].sum()
print("total revenue:", total_revenue)
```

```
product_sales = df.groupby("product") ["quantity"].sum()
print(product_sales)
best_seller = product_sales.idxmax()
print("best-selling product:", best_seller)
```

```
print(df.groupby("product") ["quantity"].sum())
```

```
print(df.groupby("product") ["quantity"].sum())
```

```
product_sales = df.groupby("product") ["quantity"].sum()
print(product_sales["phone"])
```

```
product_sales = df.groupby("product") ["quantity"].sum()
best_seller = product_sales.idxmax()
print("best-salling product:", best_seller)
```

```
product_sales.plot(kind="bar")
plt.title("quantity sold by product")
plt.xlabel("product")
plt.ylabel("units sold")
plt.show()
```

```
product_sales.sort_values().plot(kind="bar", title="total units sold by product")
plt.xlabel("product")
plt.ylabel("units sold")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
